# Run results

The results of this run folder, as they were when the notebook loaded them. The table is the scan's table; the plot is a scan's first value or the ionisation of the first point. To load new results, run the loading cell again.

```python
from pathlib import Path

import matplotlib.pyplot as plt

from srim.notebook import load_run
```

```python
# The run folder is the notebook's own folder (the working directory of the kernel).
folder = Path.cwd()
run = load_run(folder)
run.mark_loaded()
state = "finished" if run.finished else "has not finished"
print(f"{run.count} point(s) with results; the run {state}.")
```

```python
run.result.table
```

```python
if not run.points():
    print("No point has results yet: run the simulation, then run the loading cell again.")
elif len(run.result.table) > 1 and run.result.scan.extract:
    run.result.plot()
else:
    ioniz = run.point(run.points()[0]).ioniz.df
    fig, ax = plt.subplots()
    ax.plot(ioniz["depth_A"], ioniz["ions_eV_per_A_per_ion"], label="ions")
    ax.plot(ioniz["depth_A"], ioniz["recoils_eV_per_A_per_ion"], label="recoils")
    ax.set(xlabel="depth (Å)", ylabel="ionisation (eV/Å/ion)")
    ax.legend()
```
